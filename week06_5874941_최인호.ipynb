## 문제 1. 은닉층 1개 모델 (25점)
`nn.Module` 을 상속해 아래 조건의 모델을 작성하세요.

- 입력 특성 5개, 은닉층 뉴런 16개, 출력 3개
- `__init__` 과 `forward` 를 모두 구현하고 각 줄에 주석
- `print(model)` 로 구조 출력


```python
# 답안
import torch
import torch.nn as nn

# nn.Module을 상속받아 커스텀 신경망 클래스 정의
class SimpleMLP(nn.Module):
    def __init__(self):
        # 부모 클래스(nn.Module)의 __init__ 메서드를 호출하여 초기화
        super(SimpleMLP, self).__init__()

        # 입력 특성 5개 -> 은닉층 뉴런 16개로 변환하는 첫 번째 선형 레이어 정의
        self.fc1 = nn.Linear(in_features=5, out_features=16)

        # 은닉층에 적용할 활성화 함수(ReLU) 정의
        self.relu = nn.ReLU()

        # 은닉층 뉴런 16개 -> 출력 뉴런 3개로 변환하는 두 번째 선형 레이어 정의
        self.fc2 = nn.Linear(in_features=16, out_features=3)

    def forward(self, x):
        # 입력 데이터 x를 첫 번째 선형 레이어(fc1)에 통과시킴
        x = self.fc1(x)

        # 선형 연산 결과에 활성화 함수(ReLU)를 적용
        x = self.relu(x)

        # 활성화 함수를 거친 값을 출력 레이어(fc2)에 통과시킴
        x = self.fc2(x)

        # 최종 연산된 결과 반환
        return x

# 모델 객체 생성
model = SimpleMLP()

# 모델의 구조 출력
print(model)
```

## 문제 2. 은닉층 2개 모델 (25점)
문제 1과 같은 입력·출력에 은닉층을 2개로 늘린 모델을 작성하고 구조를 출력하세요.


```python
# 답안
import torch
import torch.nn as nn

# nn.Module을 상속받아 커스텀 신경망 클래스 정의
class MultiLayerMLP(nn.Module):
    def __init__(self):
        # 부모 클래스(nn.Module)의 __init__ 메서드를 호출하여 초기화
        super(MultiLayerMLP, self).__init__()

        # 입력 특성 5개 -> 첫 번째 은닉층 뉴런 16개로 변환하는 선형 레이어 정의
        self.fc1 = nn.Linear(in_features=5, out_features=16)

        # 첫 번째 은닉층 뒤에 적용할 활성화 함수(ReLU) 정의
        self.relu1 = nn.ReLU()

        # 첫 번째 은닉층 16개 -> 두 번째 은닉층 뉴런 16개로 변환하는 선형 레이어 정의
        self.fc2 = nn.Linear(in_features=16, out_features=16)

        # 두 번째 은닉층 뒤에 적용할 활성화 함수(ReLU) 정의
        self.relu2 = nn.ReLU()

        # 두 번째 은닉층 16개 -> 출력 뉴런 3개로 변환하는 선형 레이어 정의
        self.fc3 = nn.Linear(in_features=16, out_features=3)

    def forward(self, x):
        # 입력 데이터 x를 첫 번째 선형 레이어(fc1)에 통과시킴
        x = self.fc1(x)

        # 첫 번째 활성화 함수(ReLU) 적용
        x = self.relu1(x)

        # 두 번째 선형 레이어(fc2)에 통과시킴
        x = self.fc2(x)

        # 두 번째 활성화 함수(ReLU) 적용
        x = self.relu2(x)

        # 출력 레이어(fc3)에 통과시킴
        x = self.fc3(x)

        # 최종 연산된 결과 반환
        return x

# 모델 객체 생성
model = MultiLayerMLP()

# 모델의 구조 출력
print(model)
```

## 문제 3. 두 모델의 구조 비교표 (30점)
아래 항목으로 표를 만들어 출력하세요.

| 항목 | 은닉층 1개 | 은닉층 2개 |
| --- | --- | --- |
| Linear 층 수 | | |
| 전체 파라미터 수 | | |
| 활성화 함수 개수 | | |
| 출력 shape | | |


```python
# 답안 — pd.DataFrame 으로 만들어 출력하세요
import pandas as pd
import torch
import torch.nn as nn

# 1. 모델 정의
class SimpleMLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(5, 16)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(16, 3)

    def forward(self, x):
        return self.fc2(self.relu(self.fc1(x)))

class MultiLayerMLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(5, 16)
        self.relu1 = nn.ReLU()
        self.fc2 = nn.Linear(16, 16)
        self.relu2 = nn.ReLU()
        self.fc3 = nn.Linear(16, 3)

    def forward(self, x):
        return self.fc3(self.relu2(self.fc2(self.relu1(self.fc1(x)))))

# 2. 파라미터 수 계산 함수
def count_parameters(model):
    return sum(p.numel() for p in model.parameters() if p.requires_grad)

# 3. 모델 인스턴스 생성 및 데이터 추출
model1 = SimpleMLP()
model2 = MultiLayerMLP()

# 임의의 입력 데이터 (batch_size=1, feature=5)
dummy_input = torch.randn(1, 5)

data = {
    "항목": [
        "Linear 층 수",
        "전체 파라미터 수",
        "활성화 함수 개수",
        "출력 shape"
    ],
    "은닉층 1개": [
        2,
        count_parameters(model1),
        1,
        tuple(model1(dummy_input).shape)
    ],
    "은닉층 2개": [
        3,
        count_parameters(model2),
        2,
        tuple(model2(dummy_input).shape)
    ]
}

# 4. DataFrame 생성
df = pd.DataFrame(data)

# 출력
print(df)
```

## 문제 4. 구조 차이 설명 (20점)

2. 활성화 함수가 없으면 층을 쌓아도 소용없는 이유

2. 아무리 층을 쌓아도 선형으로 되어버리기 때문이다.

3.항상 좋지는 않다고 본다. 상황에 따라 층을 늘리거나 아니면 단순하게 층을 구성해야 하며 층을 단순하게 만들어 해결할 수 있는 상황일 경우 무리하게 층을 복잡하게 만들면 효율이 떨어지기 때문이다.


## 문제 5. 임의 입력으로 확인 (보너스)
두 모델에 각각 `torch.randn(8, 5)` 를 넣어 출력 shape을 확인하고, 중간 층의 shape 변화를 출력하세요.


```python
# 답안
import torch
import torch.nn as nn

# 1. 은닉층 1개 모델
class SimpleMLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(5, 16)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(16, 3)

    def forward(self, x):
        print(f"  [SimpleMLP] 입력 shape        : {x.shape}")
        x = self.fc1(x)
        print(f"  [SimpleMLP] fc1 통과 후 shape   : {x.shape}")
        x = self.relu(x)
        print(f"  [SimpleMLP] relu 통과 후 shape  : {x.shape}")
        x = self.fc2(x)
        print(f"  [SimpleMLP] fc2 (최종) shape    : {x.shape}")
        return x

# 2. 은닉층 2개 모델
class MultiLayerMLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(5, 16)
        self.relu1 = nn.ReLU()
        self.fc2 = nn.Linear(16, 16)
        self.relu2 = nn.ReLU()
        self.fc3 = nn.Linear(16, 3)

    def forward(self, x):
        print(f"  [MultiLayerMLP] 입력 shape        : {x.shape}")
        x = self.fc1(x)
        print(f"  [MultiLayerMLP] fc1 통과 후 shape   : {x.shape}")
        x = self.relu1(x)
        print(f"  [MultiLayerMLP] relu1 통과 후 shape : {x.shape}")
        x = self.fc2(x)
        print(f"  [MultiLayerMLP] fc2 통과 후 shape   : {x.shape}")
        x = self.relu2(x)
        print(f"  [MultiLayerMLP] relu2 통과 후 shape : {x.shape}")
        x = self.fc3(x)
        print(f"  [MultiLayerMLP] fc3 (최종) shape    : {x.shape}")
        return x

# 모델 인스턴스 생성
model1 = SimpleMLP()
model2 = MultiLayerMLP()

# torch.randn(8, 5) 입력 데이터 생성
input_data = torch.randn(8, 5)

print("=== 은닉층 1개 모델 (SimpleMLP) Shape 변화 ===")
out1 = model1(input_data)

print("\n=== 은닉층 2개 모델 (MultiLayerMLP) Shape 변화 ===")
out2 = model2(input_data)
```
